````python
# ============================================================================
# 📦 COMPLETE SETUP: All Imports, Functions, and Utilities
# ============================================================================

print("="*80)
print("🔧 LOADING ALL REQUIRED LIBRARIES AND UTILITIES")
print("="*80)

# Standard imports
import json
import pandas as pd
import numpy as np
import random
from collections import Counter
from typing import Dict, List, Tuple, Optional

# PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Transformers
from transformers import (
    AutoTokenizer, AutoConfig, AutoModelForSequenceClassification,
    Trainer, TrainingArguments, EarlyStoppingCallback, TrainerCallback
)

# Scikit-learn
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, precision_recall_fscore_support,
    classification_report, confusion_matrix
)
from sklearn.utils import resample

# SMOTE for data balancing
try:
    from imblearn.over_sampling import SMOTE, BorderlineSMOTE, ADASYN
    print("✓ imbalanced-learn available (SMOTE supported)")
except ImportError:
    print("⚠ imbalanced-learn not found. Install: pip install imbalanced-learn")
    SMOTE = None

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Warnings
import warnings
warnings.filterwarnings('ignore')

print("✓ All libraries imported successfully\n")

# ============================================================================
# 1. CONFIGURATION
# ============================================================================
MODEL_NAME = "microsoft/deberta-v3-large"
MAX_LENGTH = 512
NUM_CLASSES = 3

# Label mapping
label2id = {"Yes": 0, "To some extent": 1, "No": 2}
id2label = {0: "Yes", 1: "To some extent", 2: "No"}

print(f"✓ Model: {MODEL_NAME}")
print(f"✓ Max Length: {MAX_LENGTH}")
print(f"✓ Classes: {list(label2id.keys())}\n")

# ============================================================================
# 2. DATA LOADING FUNCTIONS
# ============================================================================

def extract_student_responses_from_conversation(conv_history: str) -> str:
    """Extract only student responses from conversation history."""
    student_responses = []
    lines = conv_history.split('\n')
    for line in lines:
        line = line.strip()
        if line.startswith('Student:'):
            student_text = line.replace('Student:', '').strip()
            if student_text:  # Only add non-empty responses
                student_responses.append(student_text)
    return ' '.join(student_responses)

def load_train_data_with_answers(file_path: str) -> pd.DataFrame:
    """Load training data with annotations and answer keys."""
    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    samples = []
    for conversation in data:
        conv_history = conversation['conversation_history']
        conv_id = conversation['conversation_id']
        answer_key = conversation.get('answer_key', '')
        student_responses = extract_student_responses_from_conversation(conv_history)
        
        # Skip if no student responses
        if not student_responses:
            continue
            
        for model_name, response_data in conversation['tutor_responses'].items():
            if 'annotation' in response_data and 'Mistake_Identification' in response_data['annotation']:
                samples.append({
                    'student_responses': student_responses,
                    'answer_key': answer_key,
                    'tutor_response': response_data['response'],
                    'label': response_data['annotation']['Mistake_Identification'],
                    'conversation_id': conv_id,
                    'model': model_name
                })
    
    df = pd.DataFrame(samples)
    print(f"✓ Loaded {len(df)} training samples")
    print(f"  Distribution: {dict(df['label'].value_counts())}")
    return df

print("✓ Data loading functions defined\n")

# ============================================================================
# 3. DATA AUGMENTATION FUNCTIONS
# ============================================================================

def augment_text_simple(text: str, prob: float = 0.1) -> str:
    """Simple data augmentation: random word deletion/swap."""
    words = text.split()
    if len(words) < 3:
        return text
    
    # Random deletion (10% chance per word)
    if random.random() < prob:
        idx_to_delete = random.randint(0, len(words) - 1)
        words = words[:idx_to_delete] + words[idx_to_delete + 1:]
    
    # Random swap (10% chance)
    if random.random() < prob and len(words) >= 2:
        idx1, idx2 = random.sample(range(len(words)), 2)
        words[idx1], words[idx2] = words[idx2], words[idx1]
    
    return ' '.join(words)

print("✓ Data augmentation functions defined\n")

# ============================================================================
# 4. DATA BALANCING FUNCTIONS
# ============================================================================

def balance_dataset_oversample(df: pd.DataFrame, target_col: str = 'label') -> pd.DataFrame:
    """Simple oversampling: duplicate minority classes."""
    class_counts = df[target_col].value_counts()
    max_count = class_counts.max()
    
    balanced_dfs = []
    for label in class_counts.index:
        class_df = df[df[target_col] == label]
        if len(class_df) < max_count:
            class_df_resampled = resample(
                class_df, n_samples=max_count, replace=True, random_state=42
            )
            balanced_dfs.append(class_df_resampled)
        else:
            balanced_dfs.append(class_df)
    
    balanced_df = pd.concat(balanced_dfs, ignore_index=True)
    return balanced_df.sample(frac=1, random_state=42).reset_index(drop=True)

def balance_dataset_adaptive(df: pd.DataFrame, target_col: str = 'label') -> pd.DataFrame:
    """Adaptive oversampling: balance to sqrt of max count."""
    class_counts = df[target_col].value_counts()
    max_count = class_counts.max()
    target_count = int(np.sqrt(max_count) * max(class_counts.min(), 100))
    
    balanced_dfs = []
    for label in class_counts.index:
        class_df = df[df[target_col] == label]
        if len(class_df) < target_count:
            class_df_resampled = resample(
                class_df, n_samples=target_count, replace=True, random_state=42
            )
            balanced_dfs.append(class_df_resampled)
        else:
            balanced_dfs.append(class_df)
    
    balanced_df = pd.concat(balanced_dfs, ignore_index=True)
    return balanced_df.sample(frac=1, random_state=42).reset_index(drop=True)

def balance_dataset_smote(df: pd.DataFrame, tokenizer, max_length: int = 512) -> pd.DataFrame:
    """SMOTE: Generate synthetic samples for minority classes."""
    if SMOTE is None:
        print("⚠ SMOTE not available, falling back to oversampling")
        return balance_dataset_oversample(df)
    
    # Prepare features: tokenize all texts
    texts = []
    for _, row in df.iterrows():
        if 'answer_key' in row and pd.notna(row['answer_key']) and row['answer_key']:
            text = f"{row['student_responses']} [SEP] {row['answer_key']} [SEP] {row['tutor_response']}"
        else:
            text = f"{row['student_responses']} [SEP] {row['tutor_response']}"
        texts.append(text)
    
    # Get embeddings (use mean pooling of tokens as features)
    # For SMOTE, we'll use a simpler approach: TF-IDF or count features
    from sklearn.feature_extraction.text import TfidfVectorizer
    
    vectorizer = TfidfVectorizer(max_features=100, ngram_range=(1, 2))
    X = vectorizer.fit_transform(texts).toarray()
    y = df['label'].map(label2id).values
    
    # Apply SMOTE
    smote = SMOTE(random_state=42, k_neighbors=min(5, Counter(y).most_common()[-1][1] - 1))
    X_resampled, y_resampled = smote.fit_resample(X, y)
    
    # For new synthetic samples, we need to create text (approximate)
    # We'll use the original samples and just oversample for simplicity
    # True SMOTE on embeddings would require more complex reconstruction
    print(f"  ⚠ Using adaptive oversampling instead of full SMOTE (text reconstruction complex)")
    return balance_dataset_adaptive(df)

def balance_dataset_hybrid(df: pd.DataFrame, target_col: str = 'label') -> pd.DataFrame:
    """Hybrid: Oversample minorities + undersample majority."""
    class_counts = df[target_col].value_counts()
    minority_classes = class_counts[class_counts < class_counts.max() / 2]
    
    # Target: balance between max and median
    target_count = int(class_counts.median() * 1.5)
    
    balanced_dfs = []
    for label in class_counts.index:
        class_df = df[df[target_col] == label]
        current_count = len(class_df)
        
        if current_count < target_count:
            # Oversample minorities
            class_df_resampled = resample(
                class_df, n_samples=target_count, replace=True, random_state=42
            )
            balanced_dfs.append(class_df_resampled)
        elif current_count > target_count * 2:
            # Undersample majority
            class_df_resampled = resample(
                class_df, n_samples=target_count * 2, replace=False, random_state=42
            )
            balanced_dfs.append(class_df_resampled)
        else:
            balanced_dfs.append(class_df)
    
    balanced_df = pd.concat(balanced_dfs, ignore_index=True)
    return balanced_df.sample(frac=1, random_state=42).reset_index(drop=True)

print("✓ Data balancing functions defined (oversample, adaptive, SMOTE, hybrid)\n")

# ============================================================================
# 5. DATASET CLASS
# ============================================================================

class MistakeIdentificationDataset(Dataset):
    """Enhanced dataset with optional augmentation."""
    
    def __init__(self, dataframe: pd.DataFrame, tokenizer, max_length: int = 512, 
                 has_labels: bool = True, augment: bool = False, augment_prob: float = 0.0):
        self.data = dataframe.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.has_labels = has_labels
        self.augment = augment
        self.augment_prob = augment_prob
        
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        row = self.data.iloc[idx]
        
        # Construct text
        if 'answer_key' in row and pd.notna(row['answer_key']) and row['answer_key']:
            text = f"{row['student_responses']} [SEP] {row['answer_key']} [SEP] {row['tutor_response']}"
        else:
            text = f"{row['student_responses']} [SEP] {row['tutor_response']}"
        
        # Apply augmentation (for minority classes during training)
        if self.augment and self.has_labels:
            label = row['label']
            if label in ['No', 'To some extent'] and random.random() < self.augment_prob:
                text = augment_text_simple(text, prob=0.1)
        
        # Tokenize
        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )
        
        item = {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
        }
        
        if self.has_labels:
            item['labels'] = torch.tensor(label2id[row['label']], dtype=torch.long)
        
        return item

print("✓ Dataset class defined with augmentation support\n")

# ============================================================================
# 6. FOCAL LOSS IMPLEMENTATION
# ============================================================================

class FocalLoss(nn.Module):
    """
    Focal Loss for addressing class imbalance.
    FL(p_t) = -alpha_t * (1 - p_t)^gamma * log(p_t)
    
    Args:
        alpha: Class weights (tensor of shape [num_classes])
        gamma: Focusing parameter (default 2.0)
        reduction: 'mean' or 'sum'
    """
    
    def __init__(self, alpha: Optional[torch.Tensor] = None, gamma: float = 2.0, 
                 reduction: str = 'mean'):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction
    
    def forward(self, inputs: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        """
        Args:
            inputs: Logits from model (batch_size, num_classes)
            targets: Ground truth labels (batch_size,)
        """
        # Get probabilities
        probs = F.softmax(inputs, dim=1)
        
        # Get class probabilities
        class_probs = probs.gather(1, targets.unsqueeze(1)).squeeze(1)
        
        # Compute focal weight: (1 - p_t)^gamma
        focal_weight = (1.0 - class_probs) ** self.gamma
        
        # Compute cross entropy
        ce_loss = F.cross_entropy(inputs, targets, reduction='none', weight=self.alpha)
        
        # Apply focal weight
        focal_loss = focal_weight * ce_loss
        
        if self.reduction == 'mean':
            return focal_loss.mean()
        elif self.reduction == 'sum':
            return focal_loss.sum()
        else:
            return focal_loss

print("✓ Focal Loss implemented\n")

# ============================================================================
# 7. CUSTOM TRAINER WITH FOCAL LOSS
# ============================================================================

class WeightedFocalTrainer(Trainer):
    """Custom Trainer with Focal Loss and class weights."""
    
    def __init__(self, focal_loss_fn: Optional[FocalLoss] = None, 
                 *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.focal_loss_fn = focal_loss_fn
    
    def compute_loss(self, model, inputs, return_outputs=False):
        """Override compute_loss to use focal loss."""
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")
        
        if self.focal_loss_fn is not None:
            loss = self.focal_loss_fn(logits, labels)
        else:
            # Standard cross-entropy with label smoothing
            loss = F.cross_entropy(logits, labels, label_smoothing=self.args.label_smoothing_factor)
        
        return (loss, outputs) if return_outputs else loss

print("✓ Custom Trainer defined\n")

# ============================================================================
# 8. METRICS COMPUTATION
# ============================================================================

def compute_metrics(eval_pred) -> Dict[str, float]:
    """Compute comprehensive metrics focusing on F1-macro."""
    predictions, labels = eval_pred
    preds = np.argmax(predictions, axis=1)
    
    # Overall metrics
    accuracy = accuracy_score(labels, preds)
    f1_macro = f1_score(labels, preds, average='macro', zero_division=0)
    f1_weighted = f1_score(labels, preds, average='weighted', zero_division=0)
    
    # Per-class metrics
    precision, recall, f1, support = precision_recall_fscore_support(
        labels, preds, average=None, labels=[0, 1, 2], zero_division=0
    )
    
    return {
        'accuracy': accuracy,
        'f1_macro': f1_macro,  # Primary metric!
        'f1_weighted': f1_weighted,
        'f1_yes': f1[0],
        'f1_to_some_extent': f1[1],
        'f1_no': f1[2],
        'precision_yes': precision[0],
        'precision_to_some_extent': precision[1],
        'precision_no': precision[2],
        'recall_yes': recall[0],
        'recall_to_some_extent': recall[1],
        'recall_no': recall[2],
    }

print("✓ Metrics function defined\n")

# ============================================================================
# 9. EFFECTIVE NUMBER WEIGHTS
# ============================================================================

def compute_effective_weights(labels: np.ndarray, beta: float = 0.9999) -> torch.Tensor:
    """
    Compute effective number of samples for class weights.
    More robust than inverse frequency for extreme imbalance.
    
    Paper: "Class-Balanced Loss Based on Effective Number of Samples"
    """
    class_counts = Counter(labels)
    num_classes = len(class_counts)
    
    effective_weights = {}
    for cls in range(num_classes):
        count = class_counts.get(cls, 1)
        # Effective number
        effective_num = (1.0 - beta ** count) / (1.0 - beta) if count > 0 else 1.0
        # Weight is inversely proportional to effective number
        effective_weights[cls] = (1.0 - beta) / effective_num
    
    # Normalize
    weight_sum = sum(effective_weights.values())
    for cls in effective_weights:
        effective_weights[cls] = effective_weights[cls] / weight_sum * num_classes
    
    # Convert to tensor
    weights_array = np.array([effective_weights[i] for i in range(num_classes)])
    return torch.tensor(weights_array, dtype=torch.float32)

print("✓ Effective weights computation defined\n")

# ============================================================================
# 10. LAYER-WISE LEARNING RATE
# ============================================================================

def get_optimizer_grouped_parameters(model, learning_rate: float, weight_decay: float,
                                    layerwise_lr_decay: float) -> List[Dict]:
    """
    Create parameter groups with layer-wise learning rate decay.
    Lower layers get lower learning rates.
    """
    no_decay = ["bias", "LayerNorm.weight", "LayerNorm.bias"]
    optimizer_grouped_parameters = []
    
    # Get all layers
    layers = [model.deberta.embeddings] + list(model.deberta.encoder.layer)
    layers.reverse()  # Bottom to top
    
    assigned_params = set()
    
    # Layer-wise LR for encoder
    for layer_num, layer in enumerate(layers):
        lr = learning_rate * (layerwise_lr_decay ** layer_num)
        
        # Parameters with weight decay
        params_with_decay = [
            p for n, p in layer.named_parameters()
            if not any(nd in n for nd in no_decay) and id(p) not in assigned_params and p.requires_grad
        ]
        if params_with_decay:
            optimizer_grouped_parameters.append({
                "params": params_with_decay,
                "weight_decay": weight_decay,
                "lr": lr
            })
            assigned_params.update([id(p) for p in params_with_decay])
        
        # Parameters without weight decay
        params_no_decay = [
            p for n, p in layer.named_parameters()
            if any(nd in n for nd in no_decay) and id(p) not in assigned_params and p.requires_grad
        ]
        if params_no_decay:
            optimizer_grouped_parameters.append({
                "params": params_no_decay,
                "weight_decay": 0.0,
                "lr": lr
            })
            assigned_params.update([id(p) for p in params_no_decay])
    
    # Classifier head (full learning rate)
    classifier_params_with_decay = [
        p for n, p in model.classifier.named_parameters()
        if not any(nd in n for nd in no_decay) and id(p) not in assigned_params and p.requires_grad
    ]
    if classifier_params_with_decay:
        optimizer_grouped_parameters.append({
            "params": classifier_params_with_decay,
            "weight_decay": weight_decay,
            "lr": learning_rate
        })
    
    classifier_params_no_decay = [
        p for n, p in model.classifier.named_parameters()
        if any(nd in n for nd in no_decay) and id(p) not in assigned_params and p.requires_grad
    ]
    if classifier_params_no_decay:
        optimizer_grouped_parameters.append({
            "params": classifier_params_no_decay,
            "weight_decay": 0.0,
            "lr": learning_rate
        })
    
    return optimizer_grouped_parameters

print("✓ Layer-wise LR function defined\n")

# ============================================================================
# 11. THRESHOLD OPTIMIZATION
# ============================================================================

def optimize_thresholds(y_true: np.ndarray, y_proba: np.ndarray, 
                       num_classes: int = 3) -> Tuple[np.ndarray, float]:
    """
    Find optimal decision thresholds per class to maximize F1-macro.
    
    Returns:
        best_thresholds: Optimal thresholds for each class
        best_f1_macro: F1-macro achieved with optimal thresholds
    """
    from scipy.optimize import minimize
    
    def f1_macro_with_thresholds(thresholds):
        """Compute F1-macro for given thresholds."""
        # Apply thresholds
        preds = np.argmax(y_proba - thresholds, axis=1)
        f1_macro = f1_score(y_true, preds, average='macro', zero_division=0)
        return -f1_macro  # Negative for minimization
    
    # Start with default thresholds (0 offset)
    initial_thresholds = np.zeros(num_classes)
    
    # Optimize
    result = minimize(
        f1_macro_with_thresholds,
        initial_thresholds,
        method='Nelder-Mead',
        options={'maxiter': 1000, 'xatol': 1e-4, 'fatol': 1e-4}
    )
    
    best_thresholds = result.x
    best_f1_macro = -result.fun
    
    return best_thresholds, best_f1_macro

print("✓ Threshold optimization function defined\n")

# ============================================================================
# 12. LOAD TOKENIZER
# ============================================================================
print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
print(f"✓ Tokenizer loaded: {tokenizer.__class__.__name__}\n")

print("="*80)
print("✅ SETUP COMPLETE - All utilities loaded!")
print("="*80)
print("\nNext steps:")
print("  1. Configure hyperparameters (already done in previous cell)")
print("  2. Load and balance data")
print("  3. Create model with focal loss")
print("  4. Train with optimized settings")
print("  5. Optimize thresholds")
print("="*80)
````

After training, we can further improve F1-macro by finding optimal decision thresholds for each class. By default, models use 0.5 (or argmax), but for imbalanced data, different thresholds per class can give better results.

**Expected gain**: +2-5% F1-macro

This is a post-processing step that doesn't require retraining!

````python
# ============================================================================
# 🎯 THRESHOLD OPTIMIZATION
# ============================================================================

print("="*80)
print("THRESHOLD OPTIMIZATION FOR F1-MACRO")
print("="*80)

if not OPTIMIZE_THRESHOLDS:
    print("\n⚠️  Threshold optimization is disabled (OPTIMIZE_THRESHOLDS=False)")
    print("Set OPTIMIZE_THRESHOLDS=True to enable this feature.")
else:
    print("\n[1/4] Getting predictions on validation set...")
    
    # Get predictions with probabilities
    predictions_output = trainer.predict(val_dataset)
    y_proba = torch.softmax(torch.tensor(predictions_output.predictions), dim=1).numpy()
    y_true = predictions_output.label_ids
    
    # Baseline: standard argmax
    y_pred_baseline = np.argmax(y_proba, axis=1)
    f1_baseline = f1_score(y_true, y_pred_baseline, average='macro')
    
    print(f"  ✓ Validation samples: {len(y_true)}")
    print(f"  ✓ Baseline F1-macro (argmax): {f1_baseline:.4f}")
    
    print("\n[2/4] Searching for optimal thresholds...")
    print("  This may take 1-2 minutes...")
    
    optimal_thresholds, optimal_f1 = optimize_thresholds(y_true, y_proba, NUM_CLASSES)
    
    print(f"\n[3/4] Optimal thresholds found:")
    print(f"  Yes:            {optimal_thresholds[0]:+.4f}")
    print(f"  To some extent: {optimal_thresholds[1]:+.4f}")
    print(f"  No:             {optimal_thresholds[2]:+.4f}")
    
    # Apply optimal thresholds
    y_pred_optimized = np.argmax(y_proba - optimal_thresholds, axis=1)
    
    # Compute detailed metrics
    from sklearn.metrics import classification_report
    
    print("\n[4/4] Results with optimized thresholds:")
    print(f"\n🎯 F1-Macro Comparison:")
    print(f"   Baseline (argmax):    {f1_baseline:.4f}")
    print(f"   Optimized:            {optimal_f1:.4f}")
    improvement_pct = (optimal_f1 - f1_baseline) / f1_baseline * 100
    print(f"   Improvement:          +{improvement_pct:.2f}%")
    
    print("\n📊 Detailed Classification Report (Optimized):")
    print(classification_report(
        y_true, y_pred_optimized,
        target_names=['Yes', 'To some extent', 'No'],
        digits=4
    ))
    
    # Save thresholds
    import json
    thresholds_dict = {
        'Yes': float(optimal_thresholds[0]),
        'To some extent': float(optimal_thresholds[1]),
        'No': float(optimal_thresholds[2]),
        'f1_macro_baseline': float(f1_baseline),
        'f1_macro_optimized': float(optimal_f1),
        'improvement_percent': float(improvement_pct)
    }
    
    with open(f'{output_dir}/optimal_thresholds.json', 'w') as f:
        json.dump(thresholds_dict, f, indent=2)
    
    print(f"\n💾 Optimal thresholds saved to: {output_dir}/optimal_thresholds.json")
    
    # Visualize confusion matrices
    print("\n[Bonus] Generating confusion matrices...")
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Baseline confusion matrix
    cm_baseline = confusion_matrix(y_true, y_pred_baseline)
    sns.heatmap(cm_baseline, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Yes', 'To some\nextent', 'No'],
                yticklabels=['Yes', 'To some\nextent', 'No'],
                ax=axes[0])
    axes[0].set_title(f'Baseline (F1-macro: {f1_baseline:.4f})')
    axes[0].set_ylabel('True Label')
    axes[0].set_xlabel('Predicted Label')
    
    # Optimized confusion matrix
    cm_optimized = confusion_matrix(y_true, y_pred_optimized)
    sns.heatmap(cm_optimized, annot=True, fmt='d', cmap='Greens',
                xticklabels=['Yes', 'To some\nextent', 'No'],
                yticklabels=['Yes', 'To some\nextent', 'No'],
                ax=axes[1])
    axes[1].set_title(f'Optimized Thresholds (F1-macro: {optimal_f1:.4f})')
    axes[1].set_ylabel('True Label')
    axes[1].set_xlabel('Predicted Label')
    
    plt.tight_layout()
    plt.savefig(f'{output_dir}/confusion_matrices_comparison.png', dpi=150, bbox_inches='tight')
    print(f"  ✓ Confusion matrices saved to: {output_dir}/confusion_matrices_comparison.png")
    plt.show()
    
    print("\n" + "="*80)
    print("✅ THRESHOLD OPTIMIZATION COMPLETE")
    print("="*80)
    
    if improvement_pct > 1.0:
        print(f"\n🎉 Significant improvement of +{improvement_pct:.2f}%!")
        print("Use these thresholds for final predictions.")
    else:
        print(f"\n✓ Small improvement of +{improvement_pct:.2f}%.")
        print("Standard argmax is already near-optimal.")
````
